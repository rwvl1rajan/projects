```python
print("welcome to data analyser app which analyse your data for free")
print("please choose one option from below list")
user_input= int(input("""
1. input data
2. display data summary
3. calculate factorial 
4. filter data by threshold
5. Sort data 
6. Display dataset Statistics(Return multiple values)
7. exit   
choose your option: """))
data= [1,2,3,4,5,6,7,8]
if user_input ==1:
  data=[]
  input_number=int(input('give a number'))
  while input_number != -1:
    data.append(input_number)
    input_number=int(input('give another number , press -1 to exit:   '))

elif user_input ==2:
    print("data summary is: data", data)
    print("data summary is: length of data", len(data))
    print("data summary is: sum of data", sum(data))
    print("data summary is: average of data", sum(data)/len(data))
    print("data summary is: maximum of data", max(data))
    print("data summary is: minimum of data", min(data))

elif user_input ==3:
    def factorial(n):
        if n == 0:
            return 1
        else:
            return n * factorial(n-1)
    number = int(input("Enter a number to calculate its factorial: "))
    print(f"The factorial of {number} is {factorial(number)}")

elif user_input ==4:
    threshold = int(input("Enter the threshold value: "))
    filtered_data= lambda x: x>threshold
    filtered_results= list(filter(filtered_data, data))
    print("filtered data above thrshold:", filtered_results)


elif user_input ==5:
    sorted_data= sorted(data)
    print("sorted data:-", sorted_data)

elif user_input ==6:
    def dataset_statistics(data):
        length = len(data)
        total_sum = sum(data)
        average = total_sum / length if length > 0 else 0
        maximum = max(data) if length > 0 else None
        minimum = min(data) if length > 0 else None
        return length, total_sum, average, maximum, minimum

    stats = dataset_statistics(data)
    print("Dataset Statistics:")
    print("Length:", stats[0])
    print("Sum:", stats[1])
    print("Average:", stats[2])
    print("Maximum:", stats[3])
    print("Minimum:", stats[4])

    
elif user_input ==7:
    print("Exiting the application. Goodbye!")
```
