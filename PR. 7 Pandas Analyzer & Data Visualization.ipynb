```python
import pandas as pd
import numpy as num
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
class data_analyzer:
    def __init__(self,path):
        print("welcome to the advance data analyzer","choose an option from below!")
        self.df=pd.read_csv(path)

        self.main_menu()

    def __str__(self):
        return " advance data analyzer 1.0 reporting"
    
    def main_menu(self):
        if self.df is not None:
            print('data set is now loaded')
        else:
            print('dataset is not loaded please load data before doing anything')

        user_choice= int(input("""
        1. Load_data
        2. Explore Data 
        3. perform dataframe operations
        4. handle missing data
        5. Generate descriptive statistics
        6. data visulisation
        7. Exit
        """))
        if user_choice==1:
            self.__load_data(input("give a path to the dataset"))
        elif user_choice==2:
            self.__explore_data()
        elif user_choice==3:
            self.__dataframe_ops()
        elif user_choice==4:
            self.null_handling()
        elif user_choice==5:
            self.__generate_stats()
        elif user_choice==6:
            self.__data_visualization()
        elif user_choice== 7:
            self.__exit()
        else:
            print("wrong selection choose again")
            self.main_menu()


    def __load_data(self,path):
        self.df= pd.read_csv(path)
        self.main_menu()



    def __explore_data(self):
        user_input= int(input("""
        1. show top 5 rows
        2. show bottom 5 rows
        3. show random 5 rows
        4. show rows and columns of the dataset
        5. show datatypes of the columns
        6. show basic info
        0. for main menu
        """))
        while user_input!=0:
            user_input= int(input("""
            1. show top 5 rows
            2. show bottom 5 rows
            3. show random 5 rows
            4. show rows and columns of the dataset
            5. show datatypes of the columns
            6. show basic info
            0. for main menu
            """))
            if user_input==1:
                print(self.df.head())
            elif user_input==2:
                print( self.df.tail())
            elif user_input==3:
                print(self.df.sample())
            elif user_input==4:
                print(self.df.shape)
            elif user_input==5:
                print(self.df.dtypes)
            elif user_input==6:
                print(self.df.info())
            else:
                print(" wrong input choosen")
            user_input= int(input("""
            1. show top 5 rows
            2. show bottom 5 rows
            3. show random 5 rows
            4. show rows and columns of the dataset
            5. show datatypes of the columns
            6. show basic info
            0. for main menu
            """))
        self.main_menu()


    def null_handling(self):
        print(self.df.isnull().sum())
        user_input= int(input("""
        1. drop na
        2. fill na 
        3. drop na in all the columns
        0. main menu
        """))
        while user_input!=0:
            if user_input==1:
                col= input("which column do you want to select")
                self.df.dropna(subset=[col], inplace= True)
                print("Done")
                print(f"new null list after fixing null values in column{col}")
                print(self.df.isnull().sum())

            elif user_input==2:
                col= input("which column do you want to select")

                if self.df[col].dtype== int or self.df[col].dtype== float:
                    self.df[col]=self.df[col].fillna(self.df[col].median())
                    print("Done")
                    print(f"new null list after fixing null values in column{col}")
                    print(self.df.isnull().sum())  



                else:
                    self.df[col]=self.df[col].fillna(self.df[col].mode()[0])
                    print("Done")
                    print(f"new null list after fixing null values in column{col}")
                    print(self.df.isnull().sum())

            elif user_input==3:
                self.df.dropna(inplace= True)
                print("Done all null values droped")
                print(f"new null list after fixing null values in entire dataset")
                print(self.df.isnull().sum())
            else:
                print('wrong input try agian')
            user_input= int(input("""
            1. drop na
            2. fill na 
            3. drop na in all the columns
            0. main menu
            """))
        self.main_menu()

    def __data_visualization(self):
        type_of_visualization= int(input("""" 
        what type of visualization you want
        1. single column,
        2. two columns"""))

        if type_of_visualization==1:
            col= input(f""" these are the columns which one do you want to select
            {self.df.columns}""")
            chart_type=int(input(""" What type of chart do you want
            1. boxplot
            2. countplot
            3. distribution
            4. 
            """))

            if chart_type==1:
                sns.boxplot(x=col, data= self.df)
                plt.title(f"outliers checking for column {col}")
                plt.show()

            elif chart_type==2:
                sns.countplot(x=col, data= self.df)
                plt.title(f"count checking for column {col}")
                plt.show()

            elif chart_type==2:
                sns.distplot(x=col, data= self.df)
                plt.title(f"distribution checking for column {col}")
                plt.show()
            else:
                print('wrong selection')
        elif type_of_visualization==2:
            xcol= input(f""" these are the columns which one do you want to select for x axis
            {self.df.columns}""")
            ycol= input(f""" these are the columns which one do you want to select for y axis
            {self.df.columns}""")

            chart_type=int(input(""" What type of chart do you want
            1. bar chart
            2. line plot
            3. scatter plot
            
            """))   
            if chart_type==1:
                sns.barplot(x=xcol, y= ycol, data= self.df, errorbar=None)  
                plt.show()
            elif chart_type==2:
                sns.lineplot(x=xcol, y= ycol, data= self.df, errorbar=None)  
                plt.show()
            elif chart_type==3:
                sns.scatterplot(x=xcol, y= ycol, data= self.df)       
                plt.show()                 
        self.main_menu()


    def __generate_stats(self):
        print(self.df.describe())
        self.main_menu() 

    def __dataframe_ops (self):
        u_input= int(input(""" what operations you want to perform
        1. create pivot table
        2. check and drop duplicates
        """))

        if u_input==1:
            print(self.df.columns)
            print("these are the columns which one do you want to go ahead with")
            pivot= self.df.pivot_table(index=(input('which column do you want to select for index')), 
            columns=(input('column you want to select')),
            values=(input('value you want to show in the table')))
            print(pivot)

        elif u_input==2:
            print(self.df.duplicated().sum())
            self.df= self.df.drop_duplicates()
            print("rechecking duplicates ", print(self.df.duplicated().sum()))


        self.main_menu()
        

    def __exit(self):
        print("ta daaaaaaa , byeeeeeeeee, seee uuuuuuuu")




                    


        

path ="correct_data.csv"

a= data_analyzer(path)
```

```python
class save_data(data_analyzer):
    def __init__(self,path):
        super().__init__(path)
        self.df.to_csv("correct_data.csv")
b= save_data(path)       
```
