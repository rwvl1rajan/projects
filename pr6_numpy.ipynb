```python
import numpy as np

class dataanalytics:
    def __init__(self):
        self.array = None
        self.array2 = None

    def array_creation(self):
        user_input = int(input("""What type of array do you want?
        Choose:
        1. 1D
        2. 2D
        3. 3D
        """))

        if user_input == 1:
            values = int(input("How many values do you want? "))
            self.array = np.random.randint(1, 10, values)
            print(self.array)

        else:
            columns = int(input("How many columns do you want? "))
            rows = int(input("How many rows do you want? "))

            if user_input == 2:
                self.array = np.random.randint(1, 10, (rows, columns))
            else:
                self.array = np.random.randint(
                    1, 10, (user_input, rows, columns)
                )

            print(self.array)

    def array2_creation(self):
        user_input = int(input("""What type of array do you want?
        Choose:
        1. 1D
        2. 2D
        3. 3D
        """))

        if user_input == 1:
            values = int(input("How many values do you want? "))
            self.array2 = np.random.randint(1, 10, values)
            print(self.array2)

        else:
            columns = int(input("How many columns do you want? "))
            rows = int(input("How many rows do you want? "))

            if user_input == 2:
                self.array2 = np.random.randint(1, 10, (rows, columns))
            else:
                self.array2 = np.random.randint(
                    1, 10, (user_input, rows, columns)
                )

            print(self.array2)

    def array_management(self):
        print('do you want to do some slicing, indexing on the array? choose what do you want to do')
        array_operations = int(input("""what do you want to do
        1. Slicing
        2. Split an array
        3. combine 2 arrays
        """))

        if array_operations == 1:
            if self.array is not None:
                if np.ndim(self.array) == 1:
                    start = int(input('which index to start from'))
                    end = int(input('till which index ')) + 1
                    print(self.array[start:end])
                else:
                    box_start = int(input('which index to start from for box'))
                    box_end = int(input('till which index for box')) + 1
                    row_start = int(input('which index to start from for rows'))
                    row_end = int(input('which index to end for rows')) + 1
                    column_start = int(input('which index to start from for columns'))
                    column_end = int(input('which index to end for columns')) + 1

                    if np.ndim(self.array) == 2:
                        print(self.array[row_start:row_end, column_start:column_end])
                    else:
                        print(self.array[box_start:box_end, row_start:row_end, column_start:column_end])
            else:
                print('no array found create one first')
                self.array_creation()
                self.array_management()

        elif array_operations == 2:
            if self.array is not None:
                if np.ndim(self.array) == 1:
                    print(f"length of array is this {len(self.array)} choose split accordingly")
                    desired_split = int(input("choose any number of desired split"))
                    print(np.split(self.array, desired_split))
                else:
                    print("create a 1d array others are not supported yet")
                    self.array_creation()
                    print(f"length of array is this {len(self.array)} choose split accordingly")
                    desired_split = int(input("choose any number of desired split"))
                    print(np.split(self.array, desired_split))

            else:
                print('no array found create one first')
                self.array_creation()
                self.array_management()

        elif array_operations == 3:
            if self.array is not None and self.array2 is not None:
                print(np.concatenate((self.array, self.array2)))
            else:
                self.array2_creation()
                self.array_management()

    def maths_operations(self, value):
        task = int(input("""what do you want to do
        1. sum
        2. minus
        3. multiply
        4. divide
        5. dot product
        6. matrix multiplication
        """))

        if self.array is not None:
            print('your array is this ', self.array)

            if task == 1:
                print(self.array + value)

            elif task == 2:
                print(self.array - value)

            elif task == 3:
                print(self.array * value)

            elif task == 4:
                print(self.array / value)

            elif task == 5:
                if self.array2 is not None:
                    if np.ndim(self.array) == 2 and np.ndim(self.array2) == 2:
                        print(np.dot(self.array, self.array2))
                    elif np.ndim(self.array) != 2:
                        print(" create a 2d array first")
                        self.array_creation()
                        self.maths_operations(value)
                    elif np.ndim(self.array2) != 2:
                        print(" create a 2d array first")
                        self.array2_creation()
                        self.maths_operations(value)
                else:
                    print(" create a 2d array first")
                    self.array2_creation()
                    self.maths_operations(value)

            elif task == 6:
                if self.array2 is not None:
                    if np.ndim(self.array) == 2 and np.ndim(self.array2) == 2:
                        print(np.matmul(self.array, self.array2))
                    elif np.ndim(self.array) != 2:
                        print(" create a 2d array first")
                        self.array_creation()
                        self.maths_operations(value)
                    elif np.ndim(self.array2) != 2:
                        print(" create a 2d array first")
                        self.array2_creation()
                        self.maths_operations(value)
                else:
                    print(" create a 2d array first")
                    self.array2_creation()
                    self.maths_operations(value)

        else:
            print("create an array first")
            self.array_creation()
            self.maths_operations(value)

    # NEW FUNCTION 1: Minimum, maximum, percentiles, correlation coefficient
    def statistics_operations(self):
        if self.array is not None:
            print("Minimum value:", np.min(self.array))
            print("Maximum value:", np.max(self.array))
            print("Percentiles:")
            print("25th percentile:", np.percentile(self.array, 25))
            print("50th percentile:", np.percentile(self.array, 50))
            print("75th percentile:", np.percentile(self.array, 75))

            if self.array2 is not None:
                if self.array.size == self.array2.size:
                    print(
                        "Correlation coefficient:",
                        np.corrcoef(
                            self.array.flatten(),
                            self.array2.flatten()
                        )[0, 1]
                    )
                else:
                    print("Arrays must have the same number of elements for correlation.")
            else:
                print("Create array2 to calculate correlation.")
        else:
            print("Create an array first.")

    # NEW FUNCTION 2: Mean, median, mode, sum, standard deviation, variance
    def basic_statistics(self):
        if self.array is not None:
            from statistics import mode

            print("Mean:", np.mean(self.array))
            print("Median:", np.median(self.array))
            print("Mode:", mode(self.array.flatten().tolist()))
            print("Sum:", np.sum(self.array))
            print("Standard deviation:", np.std(self.array))
            print("Variance:", np.var(self.array))
        else:
            print("Create an array first.")
```
