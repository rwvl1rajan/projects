```python
all_student_record=[]
student_id=0
all_subjects=set(['hindi'])
```

```python
print('hello students, welcome to the info place where your informations are stored')

user_choice= int(input('''what do you want to do, press any key
1. add_student'
2. display_all_students
3. update_student_information
4. delete Student
5. display subjects offered
choose one.....'''))

if user_choice==1:
  student_id= student_id+1
  student_name=input('what is your name')

  student_dob=input('give in proper format eg. day/month/ year (31/04/2008)')
  current_year= 2026
  student_age=current_year- int(student_dob[-1:-5:-1][::-1])
  student_data_unchangeable=(student_id, student_dob,student_age)

  subjects= [input('subject 1'), input('subject 2'), input('subject 3')]
  all_subjects.update(subjects)
  grades= input('total marks student got')
  student_data_dict= {str(student_id):{"Student id , dob and age ": student_data_unchangeable,'student name':student_name,
                                                  'subject':subjects,'student grades':grades}}
  all_student_record.append(student_data_dict)
elif user_choice==2:
  print('displaying all students record')
  for i in all_student_record:
    print('____ ____' *50)
    print()
    print(i)
    print('____ ____' *50)

elif user_choice==3:
  print('what do you want to edit?')
  user_input= input('''
  1. student_name
  2. student grades  ''')
  student_id=input('for which student id do you want to do the changes')-1
  if user_input==1:
    all_student_record[student_id]['student_id 5']['student name']= input('enter new name')
  elif user_input==2:
    all_student_record[student_id]['student_id 5']['student grades']= input('enter revised marks')


elif user_choice==4:
    print("You are going to delete one student.")
    id_to_remove = input("Enter student ID to remove: ")

    for i in all_student_record:
        if id_to_remove in i:
            all_student_record.remove(i)
            print("Student id {} deleted successfully!".format(id_to_remove))
            break
    else:
        print("Student ID not found.")
elif user_choice ==5:
  print(f" all subjects offered are {all_subjects}")
```
