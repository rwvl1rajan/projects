```python
print("Hello and welcome to the pattern generator and number analyser app")
print('So what do you want me to do ')
print("Press 1 for pattern genrator", "Press 2 for number analyser", 'press 3 to Exit', sep='\n')
selction= int(input("select from 1,2 or 3, selected number is - "))
print('---^---*' *20)
print('')
if selction == 1:
  print('you have selected pattern genrator ')
  print('which pattern are you intrested in',' 1. Right angled triangle', '2. Pyramid', '3. diamond pattern', sep ='\n')
  pattern_selection=int(input('select any number from 1 ,2, 3, selected number is - '))
  if pattern_selection== 1:
    print('1. Right angled triangle')
    rows= int(input('how many rows do you want'))
    for i in range(rows+1):
      print('^'*i)
  elif pattern_selection ==2:
    print('2. Pyramid')
    rows= int(input('how many rows do you want'))
    for i in range(1,rows):
      print(' '*(rows-i) +'*' *(2*i-1))
  elif pattern_selection == 3:
    print('diamond pattern')
    for i in range(1,6):
      print(' '*(6-i) +'*' *(2*i-1))
    for i in range(6,0,-1):
      print(' '*(6-i) +'*' *(2*i-1))

  else:
    print('wrong selection try again')

elif selction == 2:
  print('you have selected number analyser ')
  number1= int(input("give a starting number"))
  if number1<0 :
    print('your first number is wrong only give numbers bigger than 0')
    number1= int(input("give a starting number"))
  number2= int(input("give a ending number"))
  if number1>=number2:
    print('your second number is wrong only give numbers bigger than first number')
    number2=int(input("give a ending number"))
  for i in range(number1, number2+1):
    if i%2==0:
      print(i,' is even number')
    else:
      print(i, "is odd number")

  sum= 0
  for i in range(1,10+1):
    sum= sum+i
  print('total is -', sum)


elif selction == 3:
  print('exiting nowww.... ')

else:
  print('wrong slection you can only choose number 1, 2 or 3')
```
